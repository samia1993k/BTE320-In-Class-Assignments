## Assignment 1: Income Statement

```python
sales = float(input("Enter sales ($): "))
tax_rate = float(input("Enter tax rate (%): ")) / 100

cogs = 0.4 * sales
gross_profit = sales - cogs
taxes = gross_profit * tax_rate
net_income = gross_profit - taxes

print("\nIncome Statement")
print("Sales: $", format(sales, ",.2f"))
print("Cost of Goods Sold: $", format(cogs, ",.2f"))
print("Gross Profit: $", format(gross_profit, ",.2f"))
print("Taxes: $", format(taxes, ",.2f"))
print("Net Income: $", format(net_income, ",.2f"))
```

## Assignment 2: Inventory Management (EOQ)

```python
import math

D = float(input("Enter annual demand (units/year): "))
S = float(input("Enter ordering cost ($/order): "))
H = float(input("Enter holding cost ($/year/unit): "))

eoq = math.sqrt((2 * D * S) / H)

print("\nEconomic Order Quantity (EOQ):", round(eoq, 2))
```

## Assignment 6: Rock Paper Scissors

```python
import random

play = input("Do you want to play? (yes/no): ")

while play == "yes":
    user_move = input("Choose rock, paper, or scissors: ")
    computer_move = random.choice(["rock", "paper", "scissors"])

    print("Computer chose:", computer_move)

    if user_move == computer_move:
        print("It's a tie!")
    elif user_move == "rock" and computer_move == "scissors":
        print("You win!")
    elif user_move == "paper" and computer_move == "rock":
        print("You win!")
    elif user_move == "scissors" and computer_move == "paper":
        print("You win!")
    else:
        print("Computer wins!")

    play = input("Do you want to play again? (yes/no): ")
```

## Assignment 12: Recursive String Reversal

```python
def reverse(s):
    if len(s) == 0:
        return s
    else:
        return reverse(s[1:]) + s[0]

text = input("Enter a string: ")
print("Reversed:", reverse(text))
```

## Assignment 13: ROI Calculator

```python
def roi(cost, revenue):
    if not isinstance(cost, (int, float)) or not isinstance(revenue, (int, float)):
        return "Error: Inputs must be numbers"

    if cost <= 0:
        return "Error: Cost must be greater than 0"

    return (revenue - cost) / cost

cost = float(input("Enter cost: "))
revenue = float(input("Enter revenue: "))

result = roi(cost, revenue)
print("ROI:", result)
```

## Assignment 14: Loan Payment Calculator

```python
def loan_payment(principal, annual_rate, years):
    if principal <= 0 or annual_rate <= 0 or years <= 0:
        return "Error: All inputs must be positive"

    r = annual_rate / 100 / 12
    n = years * 12

    payment = (r * principal) / (1 - (1 + r) ** (-n))
    return payment

principal = float(input("Enter loan amount: "))
rate = float(input("Enter annual interest rate (%): "))
years = int(input("Enter number of years: "))

result = loan_payment(principal, rate, years)

if isinstance(result, float):
    print("Monthly Payment:", format(result, ".2f"))
else:
    print(result)
```

## Assignment 15: Quality Control (Exceptions)

```python
def getValues(n):
    values = []

    for i in range(n):
        while True:
            try:
                num = float(input(f"Enter value {i+1}: "))
                values.append(num)
                break
            except:
                print("Invalid input. Please enter a number.")

    return values


def calcAverage(alist):
    if len(alist) <= 3:
        return "Not enough values to calculate average"

    trimmed = alist[3:]  # ignore first 3 values
    return sum(trimmed) / len(trimmed)


n = int(input("How many values? "))
values = getValues(n)

result = calcAverage(values)

if isinstance(result, float):
    print("Average =", format(result, ".2f"))
else:
    print(result)
```

## Assignment 16: Classes (Greeting Class)

```python
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal):
        if formal:
            print(self.last + ", " + self.first)
        else:
            print(self.first)

    def __str__(self):
        return self.first + " " + self.last


# test it
person = Greeting("Samia", "Khalid")

person.display(True)   # formal
person.display(False)  # informal

print(person)
```

### Employee & Manager Class

```python
class Employee:
    def __init__(self, name, wage, hours):
        self.name = name
        self.__wage = wage   # private
        self.hours = hours

    def get_wage(self):
        return self.__wage

    def set_wage(self, wage):
        if wage >= 0:
            self.__wage = wage
        else:
            print("Invalid wage")

    def calculate_pay(self):
        return self.__wage * self.hours

    def display_info(self):
        print("Name:", self.name)
        print("Weekly Pay:", self.calculate_pay())


class Manager(Employee):
    def __init__(self, name, wage, hours, bonus):
        super().__init__(name, wage, hours)
        self.bonus = bonus

    def calculate_pay(self):
        return super().calculate_pay() + self.bonus


# Test
emp = Employee("Samia", 15, 40)
mgr = Manager("Ali", 20, 40, 200)

print("Employee:")
emp.display_info()

print("\nManager:")
mgr.display_info()
```

### Point & Rectangle (Extra)

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        return math.sqrt((self.x - other.x)**2 + (self.y - other.y)**2)


class Rectangle:
    def __init__(self, p1, p2, p3, p4):
        self.p1 = p1
        self.p2 = p2
        self.p3 = p3
        self.p4 = p4

    def width(self):
        return self.p1.distance(self.p2)

    def height(self):
        return self.p2.distance(self.p3)

    def area(self):
        return self.width() * self.height()


# Test
p1 = Point(0, 0)
p2 = Point(4, 0)
p3 = Point(4, 3)
p4 = Point(0, 3)

rect = Rectangle(p1, p2, p3, p4)

print("Width:", rect.width())
print("Height:", rect.height())
print("Area:", rect.area())
```

### Student Class (Extra)

```python
class Student:
    def __init__(self, name, number):
        self.name = name
        self.number = number
        self.courses = []

    def get_name(self):
        return self.name

    def get_courses(self):
        return self.courses

    def set_name(self, name):
        self.name = name

    def enroll(self, course):
        if course not in self.courses:
            self.courses.append(course)

    def __len__(self):
        return len(self.courses)


# Test
s = Student("Samia", 101)

s.enroll("Math")
s.enroll("Business")
s.enroll("Math")  # won't duplicate

print("Name:", s.get_name())
print("Courses:", s.get_courses())
print("Number of courses:", len(s))
```
