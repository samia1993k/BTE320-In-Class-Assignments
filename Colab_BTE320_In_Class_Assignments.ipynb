```python
import random

play = input("Do you want to play? (yes/no): ")

while play == "yes":

    user_move = input("Choose rock, paper, or scissors: ")
    computer_move = random.choice(["rock", "paper", "scissors"])

    print("Computer chose:", computer_move)

    if user_move == computer_move:
        print("It's a tie!")

    elif user_move == "rock" and computer_move == "scissors":
        print("You win!")

    elif user_move == "paper" and computer_move == "rock":
        print("You win!")

    elif user_move == "scissors" and computer_move == "paper":
        print("You win!")

    else:
        print("Computer wins!")

    play = input("Do you want to play again? (yes/no): ")

print("Thanks for playing!")
```

```python
import random

play = input("Do you want to play? (yes/no): ").strip().lower()

while play == "yes":

    user_move = input("Choose rock, paper, or scissors: ").strip().lower()
    computer_move = random.choice(["rock", "paper", "scissors"])

    print("Computer chose:", computer_move)

    if user_move not in ["rock", "paper", "scissors"]:
        print("Invalid choice. Try again.")
        continue  # goes back to the top of the while loop

    if user_move == computer_move:
        print("It's a tie!")

    elif (user_move == "rock" and computer_move == "scissors") or \
         (user_move == "paper" and computer_move == "rock") or \
         (user_move == "scissors" and computer_move == "paper"):
        print("You win!")

    else:
        print("Computer wins!")

    play = input("Do you want to play again? (yes/no): ").strip().lower()

print("Thanks for playing!")
```
